Ringkasan statistik untuk beberapa kategori:

```python
print("\nValue Counts for Role:")
print(df['Role'].value_counts())
print("\nValue Counts for Affiliation:")
print(df['Affiliation'].value_counts())
print("\nValue Counts for Power Level:")
print(df['Power Level'].value_counts())
```

Memvisualisasikan dataset:

```python
sns.set_theme(style="darkgrid")
sns.set_palette("deep")

# Visualisasi distribusi menggunakan pie chart
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Distribusi role
role_counts = df['Role'].value_counts()
axes[0].pie(role_counts, labels=role_counts.index, autopct='%1.1f%%', startangle=90)
axes[0].set_title('Distribution of Character Roles')

# Distribusi affiliation
affiliation_counts = df['Affiliation'].value_counts()
axes[1].pie(affiliation_counts, labels=affiliation_counts.index, autopct='%1.1f%%', startangle=90)
axes[1].set_title('Distribution of Affiliations')

# Distribusi power level
power_level_counts = df['Power Level'].value_counts()
axes[2].pie(power_level_counts, labels=power_level_counts.index, autopct='%1.1f%%', startangle=90)
axes[2].set_title('Distribution of Power Levels')

plt.tight_layout()
plt.savefig('section1_distributions_piechart.png')
plt.show()
```
